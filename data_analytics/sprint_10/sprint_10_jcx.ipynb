## Nivel 1: API de ejemplo

### 1.1-1.2: Obtener datos

```python
url = "https://jsonplaceholder.typicode.com/"

response = requests.get(url + "posts")

if response.ok:
    if response.text:
        posts_data = response.json()
        print(f"Código de la petición: {response.status_code}")
        print(f"Descargados {len(posts_data)} posts:")
        print(posts_data[0])
        for post in posts_data:
            print(f"{post['id']}: {post['title']}")
    else:
        print("Petición OK, pero el servidor no ha devuelto datos.")
else:
    print(f"Error en la petición. Código: {response.status_code}")
```

```python
def print_response_items(api_response):
  if api_response.ok:
    if api_response.text:
        data = api_response.json()
        print(f"Código de la petición: {api_response.status_code}")
        print(f"Descargados {len(data)} items:")
        print(data[0])
        for dicc in data:
          items = list(dicc.items())
          if len(items) >= 3:
            # items[0] és la primera tupla (ex: ('id', 1))
            # items[0][0] és la clau, parelles[0][1] és el valor
            print(f"{items[0][1]}: {items[1][1]} ({items[2][1]})")
    else:
        print("Petición OK, pero el servidor no ha devuelto datos.")

  else:
      print(f"Error en la petición. Código: {api_response.status_code}")
      print(f"Detalles: {api_response.text}")
```

```python
response = requests.get(url + "users")
print_response_items(response)
```

```python
response = requests.get(url + "todos")
print_response_items(response)
```

### 1.4: Petición POST

```python
def print_response_check_details(api_response):
  print(f"Código de estado de la petición: {api_response.status_code} - {api_response.reason}\n")

  if api_response.ok:
      print("La petición se ha ejecutado con éxito.\nEl servidor ha respondido con estos datos:")
      datos_json = api_response.json()
      print(datos_json)
      print(f"\nComprobación ID respuesta: {datos_json.get('id')}")
  else:
      print("Error en crear la publicación.")
      print(f"Detalles: {api_response.text}")
```

```python
url = "https://jsonplaceholder.typicode.com/posts"

mi_post = {
    "title": "Learning to do POSTs with Python",
    "body": "Very cool. Connecting to APIs like a pro.",
    "userId": 1  # Usuari del 1 al 10
}

response = requests.post(url, json=mi_post)
print_response_check_details(response)
```

### 1.5: Petición PATCH

```python
url = "https://jsonplaceholder.typicode.com/posts/1"

updates = {
    "title": "Updated title with PATCH in Python"
}

response = requests.patch(url, json=updates)
print_response_check_details(response)
```

### 1.6: Petición DELETE

```python
url = "https://jsonplaceholder.typicode.com/posts/1"

response = requests.delete(url)
print_response_check_details(response)
```

### 2.2: Documentación

ENDPOINTS:

- /coins/{id} — Display all the coin data including name, price, market related data, website, explorers, etc.
- /simple/price — Display data such as latest coin price, market cap and 24hr trading volume.

QUERY PARAMETERS

https://docs.coingecko.com/reference/simple-price

- vs_currencies: string. default: usd. target currency of coins, comma-separated if querying more than 1 currency. *refers to /simple/supported_vs_currencies.

- ids: string. default: Bitcoin. coins' IDs, comma-separated if querying more than 1 coin. *refers to /coins/list.

- names: string. default: Bitcoin. coins' names, comma-separated if querying more than 1 coin.

- symbols: string. default: btc. coins' symbols, comma-separated if querying more than 1 coin.

- include_tokens: enum<string>. for symbols lookups, specify all to include all matching tokens. Default top returns top-ranked tokens (by market cap or volume). Available options: top, all

- include_market_cap: boolean. include market capitalization, default: false

- include_24hr_vol: boolean. include 24hr volume, default: false

- include_24hr_change: boolean. include 24hr change percentage, default: false

- include_last_updated_at: boolean. include last updated price time in UNIX, default: false

- precision: enum<string>. decimal place for currency price value

### 3.2: CSV o JSON

Buscar recursos disponibles en CSV o JSON

```python
url_search = "https://opendata-ajuntament.barcelona.cat/data/api/action/package_search"
params_search = {"q": "concerts"}

response = requests.get(url_search, params=params_search)

if response.ok:
    data_json = response.json()
    results = data_json["result"]["results"]
    print(f"Se han encontrado {len(results)} datasets relacionados.\n")

    resource_ids = []
    nombre_datasets = []

    for dataset in results:
        for recurso in dataset["resources"]:
            if recurso.get("format", "").upper() == "CSV" and recurso.get("datastore_active") == True:
                resource_ids.append(recurso["id"])
                nombre_datasets.append(dataset["title"])

    for i in range(min(len(resource_ids), 10)):
        nombre_dataset = nombre_datasets[i]
        resource_id = resource_ids[i]

        print(f" - Dataset {i}: {nombre_dataset}")
        print(f" - Resource ID: {resource_id}")

        # Inspeccionar dataset
        print("\nConectando a la base de datos...\n")
        url_datastore = "https://opendata-ajuntament.barcelona.cat/data/api/action/datastore_search"

        params_datastore = {
            "resource_id": resource_id,
            "limit": 5
        }

        response_datastore = requests.get(url_datastore, params=params_datastore)

        if response_datastore.ok:
            datos_datastore = response_datastore.json()
            registros = datos_datastore["result"]["records"]
            total_filas = datos_datastore["result"]["total"]

            print(f"INFORMACIÓN DEL DATASET:")
            print(f" -> El dataset contiene {total_filas} registros en total.")

            print("Ejemplo del primer registro:")
            for clave, valor in registros[0].items():
                print(f"> {clave}: {valor}")
            print()
        else:
            print(f"Error en la extracción de datos: {response_datastore.status_code}")

    if len(resource_ids) < 1:
        print("Lo siento. Ninguno de los datasets tiene el DataStore activado con CSV.")

else:
    print(f"Error en la búsqueda: {response.status_code}")
```

### 3.4: Convertir a DF

Convertir los resultados obtenidos a un Dataframe.

```python
url_datastore = "https://opendata-ajuntament.barcelona.cat/data/api/action/datastore_search"
resource_id = "0f706441-b9d8-47c9-9e71-ced453810a72"

params_iniciales = {
    "resource_id": resource_id,
    "limit": 1
}

respuesta_inicial = requests.get(url_datastore, params=params_iniciales)

if respuesta_inicial.ok:
    total_registros = respuesta_inicial.json()["result"]["total"]
    print(f"El dataset tiene un total de {total_registros} registros. Iniciando descarga...")

    params_totales = {
        "resource_id": resource_id,
        "limit": total_registros
    }

    respuesta_total = requests.get(url_datastore, params=params_totales)

    if respuesta_total.ok:
        datos_json = respuesta_total.json()
        registros = datos_json["result"]["records"]

        df_descarga = pd.DataFrame(registros)

        print("Carga completada con éxito!\n")
        print(f"Dimensiones del DataFrame: {df_descarga.shape}\nMuestra del DataFrame:\n")
        display(df_descarga.head())

    else:
        print(f"Error en la descarga: {respuesta_total.status_code}")

else:
    print(f"Error en la conexión inicial: {respuesta_inicial.status_code}")
```

```python
display(df_descarga.info())
display(df_descarga[['name', 'institution_name', 'addresses_road_name', 'addresses_district_name', 'addresses_neighborhood_name', 'addresses_zip_code']])
```

### 3.5: Guardar a CSV

Guardar el dataframe en un fichero CSV.

```python
columnas = ['name', 'institution_name', 'addresses_road_name', 'addresses_district_name', 'addresses_neighborhood_name', 'addresses_zip_code']
df_final = df_descarga[columnas]

df_final.to_csv("venues_bcn.csv", index=False, encoding='utf-8-sig')

print("Fichero 'venues_bcn.csv' guardado correctamente!")
```
